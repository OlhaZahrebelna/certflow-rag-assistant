# Stage 1 — Domain & Knowledge Base Design

## 1. Fictional Company

**Company:** CertFlow Solutions  
**Industry:** B2B compliance and supplier certification services  

**Business model:**  
CertFlow helps enterprise customers verify and certify suppliers before those suppliers can participate in procurement programs.

## 2. Business Problem

CertFlow operates globally and manages hundreds of certification rules covering different regions, supplier types, risk levels, industries, and exceptions.

Certification analysts need to identify the correct policy and interpret requirements consistently. Information may be distributed across global policies, regional rules, operational procedures, validation guides, FAQs, and release notes.

The future **Certification Knowledge Assistant** will help analysts retrieve the most relevant internal guidance and answer certification-related questions using source-based information.

## 6. Supplier Categories

The initial certification model supports four supplier categories:

- **Standard Supplier** — regular supplier providing non-critical products or services.
- **Strategic Supplier** — high-value supplier important to business operations.
- **Critical Supplier** — supplier whose failure could materially disrupt customer operations.
- **Regulated Supplier** — supplier operating in an industry subject to additional compliance requirements.

A supplier may belong to more than one category.

**Example:** Strategic + Regulated

## 8. Risk Classification

Every supplier receives one of four risk levels:

- **Low**
- **Medium**
- **High**
- **Critical**

Risk classification may depend on a combination of:

**Industry + Region + Supplier Category + Financial Stability + Data Access + Operational Dependency + Compliance History**


## 9. Certification Outcomes

A certification case can result in five states:

- **CERTIFIED** — all mandatory requirements are satisfied.
- **CONDITIONALLY CERTIFIED** — approved subject to specified conditions.
- **PENDING INFORMATION** — required information is missing.
- **ESCALATED** — the analyst cannot make the decision under standard rules.
- **REJECTED** — one or more mandatory requirements failed.

## 12. Versioning and Metadata

Every document will contain metadata such as:

```yaml
document_id: CF-POL-001
title: Global Supplier Certification Policy
version: 3.2
status: Active
owner: Policy Governance
region: Global
effective_date: 2026-04-01
last_reviewed: 2026-03-15
next_review: 2027-03-15
confidentiality: Internal
```

Additional metadata fields may later include:

```yaml
document_type: Corporate Policy
industry: All
supplier_category: All
risk_level: All
language: English
```

These fields will support metadata filtering and version-aware retrieval in the RAG pipeline.


## 13. Scope of the Certification Knowledge Assistant

The Certification Knowledge Assistant is designed to help certification analysts find and interpret internal certification policies, regional requirements, validation procedures, and exception rules.

The assistant can answer questions about:

- supplier eligibility;
- certification requirements;
- regional requirements;
- industry-specific rules;
- risk-based review requirements;
- required evidence and documentation;
- exception procedures;
- escalation rules;
- certification outcomes;
- policy ownership and applicability.

### System Boundary

The assistant **does not make final certification decisions**.

Its role is to retrieve relevant internal policies and provide source-based guidance to certification analysts. Final decisions remain with authorized employees according to the certification governance model.

Not every policy will use every dimension. The applicable combination will depend on the business rule and document scope.
