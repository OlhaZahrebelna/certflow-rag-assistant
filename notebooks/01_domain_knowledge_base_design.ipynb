# Stage 1 — Domain & Knowledge Base Design

This notebook defines the business domain, system boundaries, entities, workflows, document model, and retrieval requirements for the **CertFlow RAG Assistant**.

The current project focuses on **Account Data Certification** in a master-data governance environment. The certified object is an **account record**. The assistant supports analysts by retrieving relevant internal guidance; it does not make final certification decisions or update production systems.


## 1. Fictional Organization

**Organization:** Atlas Data Services (ADS)  
**Domain:** Master Data Management, Data Quality, and Account Data Certification

ADS supports enterprise teams by validating and certifying company account records used across operational and analytical systems.

The organization maintains policies and procedures covering account identity, field validation, source hierarchy, evidence requirements, duplicate prevention, hierarchy management, QA, change control, and escalation.


## 2. Business Problem

Account data may be distributed across multiple systems and may contain conflicting, incomplete, outdated, or duplicated information.

Data Operations Analysts need to determine:

- whether the correct legal entity has been identified;
- which account fields are in scope;
- which sources are authoritative for each field;
- whether current or proposed values are supported by evidence;
- whether a possible duplicate or hierarchy issue exists;
- whether QA or Governance review is required;
- which rule version is currently applicable.

The **CertFlow RAG Assistant** is designed to help analysts retrieve the most relevant internal documentation and provide source-grounded guidance consistently and quickly.


## 3. Enterprise Systems

| System | Role |
|---|---|
| **MeridianMDM** | System of record for core account master data |
| **HorizonCRM** | Consumes certified account data for commercial workflows |
| **BeaconLake** | Receives approved data for analytics and reporting |

When the same core field differs between systems, **MeridianMDM takes precedence unless an active field rule explicitly assigns ownership elsewhere**.


## 4. Core Business Entities

| Entity | Meaning |
|---|---|
| **Account** | Master-data representation of a real-world company or organization |
| **Account Record** | The certified object stored in enterprise systems |
| **Certification Request** | Request to validate or change account data |
| **Certification Scope** | Fields that must be assessed for a full or targeted certification |
| **Field** | Individual master-data attribute such as Legal Name, Domain, or Address |
| **Evidence Source** | Source used to validate a field value |
| **Data Change** | Proposed or approved modification to an account field |
| **Duplicate Candidate** | Record that may represent the same real-world entity |
| **Parent Account** | Account representing a parent organization in the hierarchy |
| **QA Review** | Independent review for selected or risk-triggered certifications |
| **Certification Record** | Audit record containing outcome, date, processor, scope, and evidence summary |
| **Escalation** | Case requiring Governance, legal, privacy, or entity-resolution review |


## 5. Roles

| Role | Responsibility |
|---|---|
| **Requester** | Submits a certification or data-change request and supporting context |
| **Data Operations Analyst** | Performs account certification and documents the decision |
| **Account Steward** | Provides business context for account relationships and hierarchy |
| **Quality Assurance Reviewer** | Independently checks selected or risk-triggered certifications |
| **Master Data Governance Lead** | Owns policy interpretation, exceptions, field definitions, and escalations |
| **System Administrator** | Manages permissions, integrations, and technical configuration |
| **RAG Assistant** | Retrieves documentation, explains rules, identifies required sources, and suggests escalation paths |


## 6. Account Certification Workflow

The operational workflow is:

**Intake**  
↓  
**Identify the correct entity**  
↓  
**Define certification scope**  
↓  
**Gather evidence**  
↓  
**Compare and validate**  
↓  
**Record the change**  
↓  
**Duplicate and hierarchy checks**  
↓  
**Quality review**  
↓  
**Certification decision**  
↓  
**Synchronization and closure**

### Key workflow principles

- Identity must be established before certification.
- Evidence must support the specific field being certified.
- Every changed field requires a standardized change reason.
- Duplicate risks must be resolved or routed before certification is completed.
- Mandatory QA cases must pass review before the account is marked as Verified.
- Approved values are synchronized from MeridianMDM to downstream systems.


## 7. Certification Scope

A certification may be:

### Full Certification
All mandatory master-data fields are reviewed.

### Targeted Certification
Only requested fields are reviewed, but identity confirmation and duplicate screening remain mandatory.

Typical fields include:

- Account ID;
- Legal Name;
- Display Name;
- Website;
- Primary Domain;
- Headquarters Address;
- Country;
- State or Province;
- Account Status;
- Parent Account;
- Ownership Type;
- Trading Symbol;
- Exchange Code;
- Industry classification;
- source links;
- verification date;
- processor;
- change reason.


## 8. Certification Outcomes

The account-level certification process uses five primary outcomes:

| Outcome | Meaning |
|---|---|
| **Verified** | All fields in scope meet active rules and required controls are complete |
| **Pending Evidence** | Evidence is missing, insufficient, or contradictory |
| **Pending QA** | Analyst review is complete but mandatory QA is still required |
| **Rejected** | The requested change is unsupported or the record is not certifiable |
| **Escalated** | Governance, legal, privacy, or material entity-resolution review is required |

The certification flag may be set to **Yes** only when all required validation, evidence, change-reason, duplicate-screening, and QA controls are satisfied.


## 9. Evidence and Source Hierarchy

The RAG system must help analysts identify the correct source requirements for each field.

Important principles:

- source quality takes precedence over source quantity;
- evidence must support the specific field being certified;
- requester-provided values are not automatically verified evidence;
- search-result snippets and AI-generated summaries are not valid certification evidence;
- conflicting evidence must be resolved or escalated;
- the assistant must distinguish approved sources from merely convenient sources.

Source-selection logic is documented in the dedicated source hierarchy and evidence policy.


## 10. Duplicate and Entity Resolution

Duplicate prevention is a core part of certification.

Analysts may compare:

- Legal Name;
- normalized company name;
- Primary Domain;
- Headquarters Address;
- Country;
- external identifiers;
- Parent Account relationship.

Similar names alone do not prove that two records represent the same entity.

If account identity cannot be established, the record must not be marked as Verified.


## 11. Quality Review and Escalation

Independent QA is mandatory for higher-risk or structurally important changes, including selected merge, legal-name, hierarchy, bulk-update, or suspected-source-manipulation cases.

Escalation may be required for:

- unresolved entity identity;
- conflicting primary sources;
- non-standard field ownership;
- requests to depart from an active rule;
- governance-sensitive hierarchy changes;
- legal or privacy concerns;
- material duplicate-resolution issues.

The RAG Assistant may recommend an escalation path but cannot approve an exception.


## 12. Document Model and Metadata

Each knowledge-base document uses YAML front matter.

Example:

```yaml
document_id: ACD-KB-001
title: Account Data Certification Overview
document_type: policy
version: "1.0"
effective_date: 2026-04-01
last_updated: 2026-03-15
owner: Master Data Governance Team
status: active
confidentiality: internal
tags:
  - account_certification
  - master_data
  - data_quality
```

Metadata will support:

- document traceability;
- source attribution;
- metadata filtering;
- version-aware retrieval;
- active-vs-historical rule distinction;
- future retrieval evaluation.

## 14. Scope of the RAG Assistant

The assistant is designed to answer questions about:

- account certification requirements;
- field validation rules;
- authoritative systems;
- required evidence;
- source hierarchy;
- duplicate prevention;
- hierarchy changes;
- request types;
- change reasons;
- QA requirements;
- escalation rules;
- certification outcomes;
- current versus historical policy versions.

### System Boundary

The assistant **does not make final certification decisions**.

It must:

- retrieve relevant documentation;
- cite the source used;
- distinguish current rules from historical changes;
- state when evidence is insufficient;
- avoid inventing facts;
- never set a certification status;
- never modify a production record.

Final certification responsibility remains with authorized employees.


## 15. Typical Business Questions

The future RAG system should be able to answer questions such as:

1. Which system is authoritative for core account fields?
2. What should an analyst do if the legal identity of an account cannot be confirmed?
3. Can an account be marked as Verified if mandatory evidence is missing?
4. Which fields must be checked during a targeted certification?
5. When is independent QA required?
6. What should happen if two primary sources contain conflicting information?
7. How should a possible duplicate be handled before certification?
8. Who should review a material parent-account hierarchy change?
9. What evidence is required to support a field change?
10. What happens when certification succeeds but synchronization to HorizonCRM fails?
11. Can requester-provided information be treated as verified evidence?
12. When should a case be escalated to Master Data Governance?
13. Which policy version should be applied when a historical rule conflicts with an active rule?
14. What information must be stored to make a certification decision reproducible?
15. Which source types are not acceptable as certification evidence?


## 16. Rule Dimensions

Certification rules may depend on one or more dimensions:

- **Field**
- **Request Type**
- **Source Type**
- **Evidence Quality**
- **Current Value**
- **Proposed Value**
- **Entity Identity**
- **Duplicate Risk**
- **Hierarchy Impact**
- **Change Type**
- **QA Requirement**
- **Exception or Escalation Condition**
- **Document Version / Effective Date**

A simplified conceptual representation is:

**Certification Rule = Field + Evidence + Entity Context + Change Context + Control Requirements**

Not every rule uses every dimension. The applicable combination depends on the field, request, and active policy.


## 17. Retrieval Challenges

The project is intentionally designed to test realistic enterprise RAG problems:

### Similar terminology
Multiple documents may mention certification, evidence, QA, or escalation.

### Cross-document reasoning
A complete answer may require information from more than one policy or procedure.

### Version awareness
Historical policy changes must not override current active rules.

### Metadata filtering
Document status, type, effective date, and section metadata may improve retrieval.

### Source grounding
Generated answers must remain traceable to retrieved documentation.

### Insufficient evidence
The system must be able to say that the knowledge base does not support a confident answer.


## 18. Planned Evaluation Scenarios

The evaluation dataset should include different question types:

| Scenario | Example |
|---|---|
| **Direct lookup** | Which system is the source of truth for core account fields? |
| **Procedure** | What steps should be followed before an account is marked Verified? |
| **Role/ownership** | Who handles policy exceptions? |
| **Evidence** | Can an AI-generated summary be used as certification evidence? |
| **Duplicate resolution** | What happens if two records may represent the same entity? |
| **Cross-document** | Who reviews a hierarchy change and when is QA required? |
| **Version-aware** | Which rule applies if an older policy conflicts with an active version? |
| **Unanswerable** | Question whose answer is not present in the knowledge base |

These scenarios will later support retrieval metrics such as Recall@K and MRR, as well as answer-level evaluation for correctness, faithfulness, and source grounding.


## 19. Design Principles for the RAG Pipeline

The future pipeline should prioritize:

1. **Accurate retrieval over fluent guessing**
2. **Current active policy over historical text**
3. **Metadata preservation through ingestion**
4. **Traceable answers with document references**
5. **Explicit handling of insufficient evidence**
6. **Evaluation-driven selection of retrieval strategy**
7. **Human responsibility for final certification decisions**

This domain design serves as the foundation for the next stages: chunking, embeddings, retrieval, reranking, generation, and evaluation.
